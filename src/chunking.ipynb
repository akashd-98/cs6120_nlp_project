```python
import os
```

```python
from langchain_text_splitters import SentenceTransformersTokenTextSplitter
```

```python
import pdfplumber

# Extract text from a single PDF
def extract_text_from_pdf(path):
    full_text = ""
    with pdfplumber.open(path) as pdf:
        for page in pdf.pages:
            page_text = page.extract_text()
            if page_text and 'THIS PAGE INTENTIONALLY LEFT BLANK' in page_text:
                continue
            if page_text:
                full_text += page_text + "\n"
    return full_text
```

```python
pdf_directory = 'C:\\Users\\gargi\\Downloads\\NLP Project CS6120\\cs6120_nlp_project\\data\\raw'
```

```python
wendel_pdf = 'Wendel-Specifications.pdf'
wendel_text = extract_text_from_pdf(os.path.join(pdf_directory, wendel_pdf))

prof_pdf = 'prof_docs.pdf'
prof_text = extract_text_from_pdf(os.path.join(pdf_directory, prof_pdf))
```

```python
splitter = SentenceTransformersTokenTextSplitter(chunk_overlap=50, tokens_per_chunk=256)
```

```python
# get method names on sentencetransformerstokentextsplitter
dir(splitter)
```

```python
wendel_chunks = splitter.split_text(text=wendel_text)
```

```python
for i, chunk in enumerate(wendel_chunks[:3]):
    print(f"Chunk {i + 1}:")
    print(chunk)
    print("-" * 80)
```

```python
wendel_chunk_directory = 'C:\\Users\\gargi\\Downloads\\NLP Project CS6120\\cs6120_nlp_project\\data\\chunks\\sentence_transformer\\wendel_specification_chunks'
# Save the chunks to a text file
for i, chunk in enumerate(wendel_chunks):
    chunk_file_name = f"wendel-secifications_chunk_{i + 1}.txt"
    output_file_path = os.path.join(wendel_chunk_directory, chunk_file_name)
    with open(output_file_path, 'a', encoding='utf-8') as f:
        f.write(chunk)
```
