```python
import os
```

```python
from langchain_text_splitters import SentenceTransformersTokenTextSplitter
```

```python
import pdfplumber

# Extract text from a single PDF
def extract_text_from_pdf(path):
    """
    Extract text from a PDF file, ignoring pages that contain the phrase
    'THIS PAGE INTENTIONALLY LEFT BLANK'.
    Args:
        path (str): Path to the PDF file.
    Returns:
        str: Extracted text from the PDF.
    """
    full_text = ""
    with pdfplumber.open(path) as pdf:
        for page in pdf.pages:
            page_text = page.extract_text() # Extract text from the page
            # Check if the page contains the phrase 'THIS PAGE INTENTIONALLY LEFT BLANK'
            # If it does, skip this page
            if page_text and 'THIS PAGE INTENTIONALLY LEFT BLANK' in page_text: 
                continue
            if page_text:
                full_text += page_text + "\n" # Append the page text to the full text
    return full_text
```

```python
pdf_directory = '../dataset/raw'
```

```python
wendel_pdf = 'Wendel-Specifications.pdf' 
wendel_text = extract_text_from_pdf(os.path.join(pdf_directory, wendel_pdf))  # Extract text from the PDF
```

```python
print(wendel_text[:1000])
```

```python
splitter = SentenceTransformersTokenTextSplitter(chunk_overlap=50, tokens_per_chunk=256)  # Initialize the text splitter
```

```python
wendel_chunks = splitter.split_text(text=wendel_text) # Split the text into chunks
```

```python
for i, chunk in enumerate(wendel_chunks[:3]):
    print(f"Chunk {i + 1}:")
    print(chunk)
    print("-" * 80)
```

```python
len(wendel_chunks)
```

```python
wendel_chunk_directory = '../dataset/chunks/sentence_transformer/wendel_specification_chunks'
os.makedirs(wendel_chunk_directory)
# Save the chunks to a text file
for i, chunk in enumerate(wendel_chunks):
    chunk_file_name = f"wendel-secifications_chunk_{i + 1}.txt"
    output_file_path = os.path.join(wendel_chunk_directory, chunk_file_name)
    with open(output_file_path, 'a', encoding='utf-8') as f:
        f.write(chunk)
```
