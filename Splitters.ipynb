```python
from langchain_text_splitters import SentenceTransformersTokenTextSplitter

splitter = SentenceTransformersTokenTextSplitter(chunk_overlap=0)
text = 'a quick brown fox jumps over the lazy dog.\n ' \
       'lazy dog and a quick brown fox.\n' \
       'a dog is lazy and fox jumps quickly.\n' \
       'fox jumps and dog is lazy.\n' \
       'dog is lazy and fox is brown who jumps over the dog.'

count_start_and_stop_tokens = 2
text_token_count = splitter.count_tokens(text=text) - count_start_and_stop_tokens
print(text_token_count)
```

```python
# read pdf as text
from PyPDF2 import PdfReader

filepath = '/content/section_1.pdf'
reader = PdfReader(filepath)
text = ''
for page in reader.pages:
    text += page.extract_text()
```

```python
splitter = SentenceTransformersTokenTextSplitter(chunk_overlap=50, tokens_per_chunk=256)
text_token_count = splitter.count_tokens(text=text) - count_start_and_stop_tokens
print(text_token_count)
```

```python
text_chunks = splitter.split_text(text=text)

print(text_chunks[1])
```

```python
for chunks in text_chunks[:3]:
    print(f"Chunk 1: {chunks}")
    print("--"*6)
```
